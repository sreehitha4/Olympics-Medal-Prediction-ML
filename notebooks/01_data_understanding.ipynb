```python
import requests

# Olympic years used for our modeling period
olympic_years = [1988, 1992, 1996, 2000, 2004, 2008, 2012, 2016]

# World Bank indicator codes
indicators = {
    "GDP_per_capita": "NY.GDP.PCAP.CD",
    "Population": "SP.POP.TOTL",
    "Life_Expectancy": "SP.DYN.LE00.IN"
}

wdi_frames = []

for indicator_name, indicator_code in indicators.items():
    url = (
        f"https://api.worldbank.org/v2/country/all/indicator/"
        f"{indicator_code}"
        f"?date=1988:2016&format=json&per_page=20000"
    )

    response = requests.get(url)

    print(f"{indicator_name} - Status code:", response.status_code)

    data = response.json()[1]

    df = pd.DataFrame(data)

    df = df[["countryiso3code", "date", "value"]].copy()

    df["Year"] = df["date"].astype(int)
    df["NOC_WDI_Code"] = df["countryiso3code"]

    df = df[df["Year"].isin(olympic_years)]

    df = df.rename(columns={"value": indicator_name})

    df = df[["NOC_WDI_Code", "Year", indicator_name]]

    wdi_frames.append(df)

wdi_data = wdi_frames[0]

for df in wdi_frames[1:]:
    wdi_data = wdi_data.merge(
        df,
        on=["NOC_WDI_Code", "Year"],
        how="outer"
    )

print("\nWDI data shape:", wdi_data.shape)
print("\nColumns:")
print(wdi_data.columns.tolist())

print("\nSample:")
print(wdi_data.head(15))
```

```python
# Fetch World Development Indicators correctly

olympic_years = [1988, 1992, 1996, 2000, 2004, 2008, 2012, 2016]

indicators = {
    "GDP_per_capita": "NY.GDP.PCAP.CD",
    "Population": "SP.POP.TOTL",
    "Life_Expectancy": "SP.DYN.LE00.IN"
}

wdi_frames = []

for indicator_name, indicator_code in indicators.items():

    all_records = []

    # Fetch all pages from the World Bank API
    first_url = (
        f"https://api.worldbank.org/v2/country/all/indicator/"
        f"{indicator_code}"
        f"?date=1988:2016&format=json&per_page=1000&page=1"
    )

    first_response = requests.get(first_url)
    first_json = first_response.json()

    total_pages = first_json[0]["pages"]

    all_records.extend(first_json[1])

    for page in range(2, total_pages + 1):
        url = (
            f"https://api.worldbank.org/v2/country/all/indicator/"
            f"{indicator_code}"
            f"?date=1988:2016&format=json&per_page=1000&page={page}"
        )

        response = requests.get(url)
        page_data = response.json()

        if len(page_data) > 1:
            all_records.extend(page_data[1])

    df = pd.DataFrame(all_records)

    # Keep only the required fields
    df = df[
        ["countryiso3code", "date", "value"]
    ].copy()

    df["Year"] = df["date"].astype(int)

    # Keep only Olympic years
    df = df[df["Year"].isin(olympic_years)]

    df = df.rename(
        columns={
            "countryiso3code": "WDI_Code",
            "value": indicator_name
        }
    )

    df = df[
        ["WDI_Code", "Year", indicator_name]
    ]

    wdi_frames.append(df)

    print(
        indicator_name,
        "records:",
        len(df)
    )

# Merge the three indicators
wdi_data = wdi_frames[0]

for df in wdi_frames[1:]:
    wdi_data = wdi_data.merge(
        df,
        on=["WDI_Code", "Year"],
        how="outer"
    )

print("\nFinal WDI shape:", wdi_data.shape)

print("\nColumns:")
print(wdi_data.columns.tolist())

print("\nSample:")
print(wdi_data.head(15))
```

```python
print("Number of non-missing WDI codes:",
      wdi_data["WDI_Code"].notna().sum())

print("\nNumber of unique WDI codes:",
      wdi_data["WDI_Code"].nunique())

print("\nFirst 20 unique WDI codes:")
print(wdi_data["WDI_Code"].dropna().unique()[:20])

print("\nNumber of blank WDI codes:")
print((wdi_data["WDI_Code"] == "").sum())
```

```python
# Remove records without a valid World Bank country code

wdi_data_clean = wdi_data[
    wdi_data["WDI_Code"].notna()
    & (wdi_data["WDI_Code"] != "")
].copy()

print("Original WDI shape:", wdi_data.shape)
print("After removing blank codes:", wdi_data_clean.shape)

print("\nNumber of unique WDI codes:",
      wdi_data_clean["WDI_Code"].nunique())

print("\nFirst 20 WDI codes:")
print(wdi_data_clean["WDI_Code"].unique()[:20])

print("\nSample:")
print(wdi_data_clean.head(15))
```

```python
# Identify WDI codes that are not directly represented by Olympic NOCs

olympic_noc_codes = set(noc_country_map["NOC"])

wdi_codes = set(wdi_data_clean["WDI_Code"])

non_olympic_codes = sorted(wdi_codes - olympic_noc_codes)

print("Number of WDI codes not matching Olympic NOCs:",
      len(non_olympic_codes))

print("\nFirst 50 non-matching codes:")
print(non_olympic_codes[:50])
```

```python
# Check how many Olympic NOCs directly match World Bank ISO-3 codes

olympic_nocs = set(noc_country_map["NOC"].dropna())
wdi_codes = set(wdi_data_clean["WDI_Code"].dropna())

direct_matches = sorted(olympic_nocs & wdi_codes)
no_direct_match = sorted(olympic_nocs - wdi_codes)

print("Olympic NOCs:", len(olympic_nocs))
print("Direct WDI matches:", len(direct_matches))
print("Olympic NOCs without direct WDI match:", len(no_direct_match))

print("\nFirst 30 direct matches:")
print(direct_matches[:30])

print("\nOlympic NOCs without direct WDI match:")
print(no_direct_match)
```

```python
# NOCs actually appearing in our modeling period (1988-2016)

modeling_period_data = complete_medal_data[
    complete_medal_data["Year"].between(1988, 2016)
].copy()

modeling_nocs = sorted(
    modeling_period_data["NOC"].dropna().unique()
)

modeling_noc_set = set(modeling_nocs)

unmatched_modeling_nocs = sorted(
    modeling_noc_set - wdi_codes
)

print("NOCs appearing in 1988-2016:", len(modeling_nocs))
print("Directly matched to WDI:", 
      len(modeling_noc_set & wdi_codes))
print("Unmatched NOCs needing mapping:",
      len(unmatched_modeling_nocs))

print("\nUnmatched NOCs in 1988-2016:")
print(unmatched_modeling_nocs)
```

```python
# Check unmatched NOCs by Olympic year

unmatched_by_year = (
    modeling_period_data[
        modeling_period_data["NOC"].isin(unmatched_modeling_nocs)
    ]
    .groupby("Year")["NOC"]
    .nunique()
    .reset_index(name="Unmatched_NOC_Count")
)

print(unmatched_by_year)

print("\nUnmatched NOCs appearing in each Olympic year:")

for year in sorted(modeling_period_data["Year"].unique()):
    year_nocs = set(
        modeling_period_data[
            modeling_period_data["Year"] == year
        ]["NOC"]
    )

    unmatched = sorted(year_nocs & set(unmatched_modeling_nocs))

    print(f"{year}: {unmatched}")
```
